```python
from smolagents import CodeAgent, tool, OpenAIModel
from google.colab import userdata

model = OpenAIModel(
    model_id="openai/gpt-oss-120b",
    api_base="https://api.groq.com/openai/v1",
    api_key=userdata.get("GROQ_API_KEY")
)
```

```python
@tool
def calculate_shipping_cost(weight: float, rate: float) -> float:
    """Calculates the total shipping cost.

    Args:
        weight: Package weight in kilograms.
        rate: Shipping rate per kilogram.
    """
    return weight * rate

agent = CodeAgent(
    tools=[calculate_shipping_cost],
    model=model,
    max_steps=5
)

agent.run(
    "Calculate the shipping cost of a 15 kg package "
    "with a rate of 3500 Colombian pesos per kilogram."
)
```
